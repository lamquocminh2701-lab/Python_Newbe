```python
n = int(input("nhap n vao day: "))
#dem các số từ 1 đến n
lst = list(range(1, n+1))
dem = len(lst)
print(lst)
print(f"đếm các số từ 1 den {n} là: {dem}")
#tính tổng các số từ 1 đến n
tong = sum(lst)
print(f"tổng các số từ 1 đên n là: {tong}")
```

```python
#nhap số chẵn lẻ
nhap = input("nhap vào một chuỗi số: ")
tach = nhap.split()
#chuyển các ký tự thành số
num = []
for x in tach:
  num.append(int(x))
#chia ra số chắn và số lẻ
chan = []
le = []
for x in num:
  if x % 2 == 0:
    chan.append(x)
  else:
    le.append(x)
#hiển thị
print(f"danh sách các số nguyên là: {num}")
print(f"danh sách các số chẵn của dãy là: {chan}")
print(f"danh sách các số lẻ của dãy là: {le}")
```

```python
#tìm số lớn nhất không cần max() cách 1
nhap = input("nhap day so cua ban vao: ")
tach = nhap.split()
#chuyen thanh num
num = []
for x in tach:
  num.append(int(x))
#sắp sếp số theo thứ tự bé đến lớn
num.sort()
#in ra số lớn nhất mà không cần dùng max()
print(f"số lớn nhất của dãy: {tach} là: {num[-1]}")
```

```python
#tìm số lớn nhất không dùng max() và sắp xếp
nhap = input("nhap day so cua ban vao danh sach: ")
#tach so thanh cac ký tự để so sánh
tach = nhap.split()
#chuyển thành số
num = []
for x in tach:
  num.append(int(x))
#tìm số lớn nhất không dùng max() hay sắp xếp
lon_nhat = num[0]
for x in num:
  if x > lon_nhat:
    lon_nhat = x
print(f"số lớn nhất trong dãy là: {lon_nhat}")
```
